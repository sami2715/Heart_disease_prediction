### Step 1: Import Libraries
---
* **Objective:** Load all necessary Python libraries required for data manipulation, numerical processing, machine learning modeling, visualization, and artifact serialization.
* **Key Modules:**
  * `pandas` & `numpy`: Data manipulation, vector operations, and tabular data structures.
  * `scikit-learn`: Data splitting (`train_test_split`), feature scaling (`StandardScaler`), modeling (`SVC`), and evaluation metrics (`classification_report`, `confusion_matrix`).
  * `matplotlib` & `seaborn`: Graphical visualization of performance matrices.
  * `pickle`: Model and scaler persistence for production deployment..

```python
# Step 1: Import all necessary libraries
import numpy as np
import pandas as pd
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
from prettytable import PrettyTable
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler  
from sklearn import svm
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

### Step 2: Load Sample Data

---

**Objective:** Read the heart disease clinical dataset (`heart.csv`) into a Pandas DataFrame.

**Key Operations:**
- Inspect row counts, column structure, and statistical summaries.
- **Target Column:** `target` (0 = No Heart Disease, 1 = Heart Disease).
- **Selected Clinical Features:**
  - `cp` — Chest Pain Type
  - `chol` — Cholesterol
  - `thalach` — Maximum Heart Rate
  - `thal` — Thalassemia

```python
"""
Purpose of this section:
-------------------------
This part of the code loads the heart disease dataset from a CSV file into a
pandas DataFrame. It then displays the dataset so that students can
see the raw data being used for further processing and model training.
"""

# Read the CSV file and load it into a pandas DataFrame
# 'heart.csv' should be present in the working directory
heart_data = pd.read_csv("heart.csv")

# Print a header to indicate that sample data is being displayed
print("\n\nSample Data:")
print("============\n")

# Configure pandas to display all rows and columns (so nothing gets hidden with '...')
pd.set_option("display.max_rows", None, "display.max_columns", None)

# Print the complete dataset
print(heart_data)
```

### Step 3: Understand and Preprocess Sample Data
---

#### Step 3.1: Understand Sample Data
* **Objective:** Check dataset dimensions, data types, missing values (`null` / `NaN`), and distribution of target labels.

---

```python
"""
Purpose of this section:
-------------------------
This part of the code helps us understand the dataset by:
1. Printing the names of all attributes (columns) in the DataFrame.
2. Showing the total number of instances (rows) available in the dataset.
"""

# Print a header for clarity
print("\n\nAttributes in Heart Disease Data:")
print("=================================\n")

# Display all column names (attributes) in the dataset
print(heart_data.columns)

# Print a header for instance count (using 'target' column to count rows)
print("\n\nNumber of Instances in Heart Disease Data:", heart_data["target"].count())
print("=============================================\n")
```

#### Step 3.2: Process Sample Data
* **Status:** **Already Completed in `heart.csv`**
* **Reasoning:** Unlike the raw Titanic dataset (which required handling missing values in `Age` or `Cabin` and dropping irrelevant text columns), `heart.csv` is a clean, pre-cleaned benchmark medical dataset. All records contain complete clinical values without missing entries.

### Step 4: Feature Extraction
---
* **Status:** **Already Completed for This Project**
* **Reasoning:** In text or complex tabular datasets (like Titanic's `Name` or `Ticket` columns), custom feature extraction is needed to derive titles or family sizes. For `heart.csv`, feature selection is driven by clinical domain knowledge: the key medical attributes (`cp`, `chol`, `thalach`, `thal`) are already extracted into structured, independent numerical variables.

### Step 5: Label Encoding the Sample Data
---
* **Status:** **Not Needed / Skipped for Heart Disease Dataset**
* **Reasoning & Adaptation:**
  * **Titanic Workflow:** Used `LabelEncoder` to translate text categorical columns (`Gender`: `'male'`/`'female'`, `Embarked`: `'Cherbourg'`/`'Southampton'`) into integers ($0, 1, 2$).
  * **Heart Disease Workflow:** All features in `heart.csv` are already purely numeric (`cp` $= 0-3$, `chol` $= \text{mg/dl}$, `thalach` $= \text{bpm}$, `thal` $= 0-3$).
  * **Alternative Requirement (Feature Scaling):** Because Support Vector Machines calculate geometric distances between data points, features with large scales (like cholesterol $\approx 250$) would dominate small features (like chest pain $\approx 2$). Instead of Label Encoding, we apply **`StandardScaler`** during Step 6 to standardize features to a zero mean ($\mu = 0$) and unit variance ($\sigma = 1$).

### Step 6: Execute the Training Phase

#### Step 6.1: Splitting Sample Data into Training and Testing Data

- Splits the dataset into an **80% Training Set** (for model learning) and a **20% Testing Set** (for evaluation) using `train_test_split` with `random_state=42` and `shuffle=True`.

```python
"""
Purpose of this section:
-------------------------
This part of the code splits the heart disease dataset into Training Data and Testing Data.
- Training Data is used to teach the machine learning model.
- Testing Data is used to check how well the model performs on unseen data.
"""

# Split dataset into training (80%) and testing (20%)
# shuffle=True (default) ensures data is randomized before splitting
training_data, testing_data = train_test_split(
    heart_data,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

# Save the training and testing datasets to CSV files
training_data.to_csv(r'training-data.csv', index=False, header=True)
testing_data.to_csv(r'testing-data.csv', index=False, header=True)

print("Training Data shape:", training_data.shape)
print("Testing Data shape:", testing_data.shape)
```

#### Step 6.2: Splitting Input Vectors and Output Labels of Training Data
* Isolates input feature columns ($X_{\text{train}}$: `cp`, `chol`, `thalach`, `thal`) from the target column ($y_{\text{train}}$: `target`).

```python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (input vectors) and
the labels (output values) from the training dataset.
"""

# Extract input vectors (features) from training data
input_vector_train = training_data.iloc[:, :-1]

# Extract output labels (target variable) from training data
output_label_train = training_data.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
```

#### Step 6.3: Train the Support Vector Classifier (SVC)
* Fits a `StandardScaler` on $X_{\text{train}}$ and transforms input vectors into scaled Z-scores ($X_{\text{train\_scaled}}$).
* Initializes and fits the `SVC(kernel='rbf')` model on scaled training vectors and target labels.

```python
"""
Purpose of this section:
-------------------------
This part of the code trains a Support Vector Classifier (SVC) on the heart disease training data.
Features are scaled using StandardScaler so large feature values (e.g., cholesterol) 
do not distort the model's decision boundary.`
"""

from sklearn.preprocessing import StandardScaler

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

# Scale feature inputs so all attributes have mean = 0 and variance = 1
scaler = StandardScaler()
input_vector_train_scaled = scaler.fit_transform(input_vector_train)

# Initialize the Support Vector Classifier with standard parameters
svc_model = svm.SVC(gamma='auto', random_state=42)

# Train the model using the scaled features and flattened labels
svc_model.fit(input_vector_train_scaled, np.ravel(output_label_train))

# Print trained model details
print(svc_model)
```

#### Step 6.4: Save the Trained Model & Scaler
* Serializes both trained artifacts (`svc_trained_model.pkl` and `scaler.pkl`) to disk using `pickle.dump()`.

```python
"""
Purpose of this section:
-------------------------
This part of the code saves both the trained Support Vector Classifier (SVC)
model and the fitted StandardScaler to your computer so they can be reused 
later without retraining.
"""

# Save the trained SVC model to disk
pickle.dump(svc_model, open('svc_trained_model.pkl', 'wb'))
print("Saved: svc_trained_model.pkl")

# Save the fitted scaler to disk (crucial for scaling future patient inputs)
pickle.dump(scaler, open('scaler.pkl', 'wb'))
print("Saved: scaler.pkl")
```

### Step 7: Execute the Testing Phase
---

#### Step 7.1: Splitting Input Vectors and Output Labels of Testing Data
* Isolates testing input features ($X_{\text{test}}$) and true target outcomes ($y_{\text{test}}$).

```python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (inputs) and the labels (outputs)
from the testing dataset.
"""

input_vector_test = testing_data.iloc[:, :-1]
output_label_test = testing_data.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
```

#### Step 7.2: Load the Saved Model
* Reloads `svc_trained_model.pkl` and `scaler.pkl` from disk to verify artifact integrity

```python
"""
Purpose of this section:
-------------------------
This part of the code loads the previously saved Support Vector Classifier (SVC)
model and the StandardScaler from disk into memory.
"""

# Load trained SVC model from disk
model = pickle.load(open('svc_trained_model.pkl', 'rb'))
print("Model loaded from svc_trained_model.pkl")

# Load fitted StandardScaler from disk
scaler = pickle.load(open('scaler.pkl', 'rb'))
print("Scaler loaded from scaler.pkl")
```

#### Step 7.3: Evaluate the Machine Learning Model
##### Step 7.3.1: Make Predictions with the Trained Model on Testing Data
* Scales test features using `scaler.transform(X_test)` and generates predictions using `model.predict()`.

```python
"""
Purpose of this section:
-------------------------
This part of the code evaluates the trained Support Vector Classifier (SVC)
on the heart disease testing data using scaled features.
"""

# Scale test features (converts DataFrame to scaled NumPy array)
input_vector_test_scaled = scaler.transform(input_vector_test)

# Predict targets on SCALED test features
model_predictions = model.predict(input_vector_test_scaled)

# Suppress pandas chained assignment warnings
pd.options.mode.chained_assignment = None

# Append predictions to 'testing_data' (not testing_data_encoded)
testing_data["Predictions"] = model_predictions

# Save full results and predictions to a CSV file
testing_data.to_csv(r'model-predictions.csv', index=False, header=True)

# Print full prediction DataFrame
print("\n\nPredictions Returned by svc_trained_model:")
print("==========================================\n")
print(testing_data)
```

#### Step 7.4: Calculate the Accuracy Score & Export Predictions
* Appends model predictions directly to the test DataFrame and exports the output to `model-predictions.csv`.

```python
"""
Purpose of this section:
-------------------------
Evaluates the SVM model on heart disease test data using:
- Classification Report: Precision, Recall, and F1-Score for each class.
- Confusion Matrix: Grid showing correct vs. incorrect diagnoses.
- Heatmap Visualization: Graphical representation of the confusion matrix.
"""


# Generate classification report
print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    testing_data["target"],
    testing_data["Predictions"],
    target_names=["No Heart Disease (0)", "Heart Disease (1)"]
))

# Generate confusion matrix table
conf_matrix = confusion_matrix(testing_data["target"], testing_data["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No Heart Disease", "Actual: Heart Disease"],
    columns=["Predicted: No Heart Disease", "Predicted: Heart Disease"]
)

print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

# Plot confusion matrix heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(
    conf_matrix, 
    annot=True, 
    fmt='d', 
    cmap='Blues',
    xticklabels=['No Heart Disease', 'Heart Disease'],
    yticklabels=['No Heart Disease', 'Heart Disease']
)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - SVC on Heart Disease Data')
plt.tight_layout()
plt.show()
```

#### Step 7.5: Detailed Evaluation Metrics
* Generates a **Classification Report** (Precision, Recall, F1-score) and visualizes a **Confusion Matrix Heatmap** to assess diagnostic accuracy and false negative rates.

```python
"""
Purpose of this section:
-------------------------
This part of the code gathers clinical input data for a single patient
(Chest Pain, Cholesterol, Max Heart Rate, Thalassemia) to predict heart disease.

To enter custom values interactively during a live notebook session, 
replace the four demo variables below with:
    cp_input = int(input("Enter Chest Pain Type (0-3): ").strip())
    chol_input = float(input("Enter Serum Cholesterol in mg/dl (e.g., 233): ").strip())
    thalach_input = float(input("Enter Maximum Heart Rate (e.g., 150): ").strip())
    thal_input = int(input("Enter Thalassemia Type (0-3): ").strip())
"""

# Demo patient medical features
cp_input = 2          # Chest Pain Type (0, 1, 2, or 3)
chol_input = 233.0    # Serum Cholesterol in mg/dl
thalach_input = 150.0 # Maximum Heart Rate achieved
thal_input = 2        # Thalassemia (0, 1, 2, or 3)

# Display summary of entered patient data
print(f"Patient Data -> Chest Pain (cp): {cp_input}, Cholesterol: {chol_input} mg/dl, Max Heart Rate: {thalach_input} bpm, Thalassemia (thal): {thal_input}")
```

### Step 8: Execute the Application Phase
---

#### Step 8.1: Take Input from User
* Prompts or assigns patient clinical values: Chest Pain Type (`cp`), Serum Cholesterol (`chol`), Maximum Heart Rate (`thalach`), and Thalassemia (`thal`).

#### Step 8.2: Convert User Input into Feature Vector
* Formats raw user input into a single-row Pandas DataFrame matching the exact feature column schema of the training set.

```python
"""
Purpose of this section:
-------------------------
This part of the code takes the patient's clinical inputs and 
converts them into a structured feature vector (DataFrame) 
matching the feature columns of the heart disease dataset.
"""

# Create a single-row DataFrame using heart disease feature names
user_input = pd.DataFrame({
    'cp': [cp_input],
    'chol': [chol_input],
    'thalach': [thalach_input],
    'thal': [thal_input]
})

print("\n\nUser Input Feature Vector:")
print("==========================\n")
print(user_input)
```

#### Step 8.3: Label Encoding of Feature Vectors
* **Status:** **Not Needed (Replaced by Scaling)**
* **Reasoning:** Since inputs are already numeric, label encoding is skipped. Instead, the raw vector is transformed via `scaler.transform(user_input)` to match the standardized feature distribution expected by the SVM.

#### Step 8.4: Load the Saved Model
* Loads the production model (`svc_trained_model.pkl`) and scaler (`scaler.pkl`) into active memory.

```python
"""
Purpose of this section:
-------------------------
This part of the code loads the previously saved Support Vector Classifier (SVC)
model and the StandardScaler from disk into memory for the Application Phase.
"""

# Load the trained Support Vector Classifier model
model = pickle.load(open('svc_trained_model.pkl', 'rb'))
print("Model loaded from svc_trained_model.pkl")

# Load the fitted StandardScaler
scaler = pickle.load(open('scaler.pkl', 'rb'))
print("Scaler loaded from scaler.pkl")
```

#### Step 8.5: Model Prediction
##### Step 8.5.1: Apply Model on Scaled Features of Unseen Instances & Return Predictions to User
* Feeds the scaled single-patient vector into `model.predict()`, maps the numeric prediction ($0$ or $1$) to clinical text (`"HEART DISEASE DETECTED"` vs. `"NO HEART DISEASE DETECTED"`), and prints a formatted output table.

```python
"""
Purpose of this section:
-------------------------
This part of the code scales the patient feature vector using the loaded 
scaler, then uses the trained Support Vector Classifier (SVC) to predict 
whether the patient has heart disease.
"""


# 1. Scale the patient features using the loaded scaler object
unseen_data_features_scaled = scaler.transform(user_input)

# 2. Predict using the SCALED feature vector
predicted_diagnosis = model.predict(unseen_data_features_scaled)

# 3. Map predicted numeric class (0 or 1) to clinical text output
if predicted_diagnosis[0] == 1:
    prediction = "HEART DISEASE DETECTED"
else:
    prediction = "NO HEART DISEASE DETECTED"

# 4. Create formatted ASCII table output
pretty_table = PrettyTable()
pretty_table.add_column("        ** Diagnosis Prediction **        ", [prediction])

print(pretty_table)
```
